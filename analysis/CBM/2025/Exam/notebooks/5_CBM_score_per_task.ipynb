```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

root_path = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "utils").exists()), None)
if root_path is None:
    raise FileNotFoundError("Could not locate project root containing 'utils' directory")
sys.path.append(str(root_path))

from utils.helpers import load_data
from utils.CBM_score_per_task import CBM_score_per_task_exam
```

```python
df = load_data("../data/CBM_format/IN1020_2025_CBM.csv")
df = df[['student_id', 'question_number', 'answer_alternative', 'confidence_level', 'auto_score', 'total_score']]
df.head()
```

### Task 2.1

```python
result2_1 = CBM_score_per_task_exam(df, question_number='2.1', map_answers=['F', 'R', 'P', 'M', 'P'])
result2_1.head()
```

```python

student_scores = result2_1.drop_duplicates(subset=["student_id"])[["student_id", "score"]]

print(f"Number of unique students: {len(student_scores)}")

#By using value_counts().sort_index(), each bar is drawn at the exact score value that students actually received
score_counts = student_scores["score"].value_counts().sort_index()

plt.figure(figsize=(10, 6))
plt.bar(score_counts.index, score_counts.values, width=0.4, edgecolor="black", align="center")
plt.xticks(score_counts.index, rotation=45, ha="right")
plt.xlabel("CBM total Score Exam")
plt.ylabel("Number of Students")
plt.title("CBM 2.1 total Scores per Student")
plt.xlim(-18, 10)
plt.ylim(0, 410)
plt.tight_layout()

#save plot to output directory
output_dir = root_path / "analysis" / "CBM" / "2025" / "Exam" / "output" / "output_task_per_task"
output_dir.mkdir(parents=True, exist_ok=True)
plt.savefig(output_dir / "CBM_2_1_score_distribution.png")

plt.show()
```

```python
#rename column score to score_2_1
result2_1 = result2_1.rename(columns={"score": "score_2_1"})
result2_1.head()
```
