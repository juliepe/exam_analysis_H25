```python
from pathlib import Path
import sys

root_path = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "utils").exists()), None)
if root_path is None:
    raise FileNotFoundError("Could not locate project root containing 'utils' directory")
sys.path.append(str(root_path))

from utils.helpers import load_data, save_excel_file
from utils.CBM_score_per_task import CBM_score_per_task_exam, CBM_score_per_task_exam_no_minus_points
```

```python
#Create a table with all the scores for each question, merge on student_id
overview_table = result2_1.merge(result2_2, on='student_id', how='outer') \
    .merge(result2_3, on='student_id', how='outer') \
    .merge(result2_4, on='student_id', how='outer') \
    .merge(result2_5, on='student_id', how='outer') \
    .merge(result2_6, on='student_id', how='outer') \
    .merge(result2_7, on='student_id', how='outer') 
#Add new column with the total score for each student
overview_table['total_CBM_score_with_minus_points'] = overview_table[['score_2_1', 'score_2_2', 'score_2_3', 'score_2_4', 'score_2_5', 'score_2_6', 'score_2_7']].sum(axis=1)
overview_table = overview_table.drop(columns=['score_2_1', 'score_2_2', 'score_2_3', 'score_2_4', 'score_2_5', 'score_2_6', 'score_2_7'])

overview_table = overview_table.merge(result_no_minus_points_2_1, on='student_id', how='outer') \
    .merge(result_no_minus_points_2_2, on='student_id', how='outer') \
    .merge(result_no_minus_points_2_3, on='student_id', how='outer') \
    .merge(result_no_minus_points_2_4, on='student_id', how='outer') \
    .merge(result_no_minus_points_2_5, on='student_id', how='outer') \
    .merge(result_no_minus_points_2_6, on='student_id', how='outer') \
    .merge(result_no_minus_points_2_7, on='student_id', how='outer') 
#Add new column with the total score for each student
overview_table['total_CBM_score_without_minus_points'] = overview_table[['score_no_minus_points_2_1', 'score_no_minus_points_2_2', 'score_no_minus_points_2_3', 'score_no_minus_points_2_4', 'score_no_minus_points_2_5', 'score_no_minus_points_2_6', 'score_no_minus_points_2_7']].sum(axis=1)
overview_table = overview_table.drop(columns=['score_no_minus_points_2_1', 'score_no_minus_points_2_2', 'score_no_minus_points_2_3', 'score_no_minus_points_2_4', 'score_no_minus_points_2_5', 'score_no_minus_points_2_6', 'score_no_minus_points_2_7'])

overview_table = overview_table.merge(df[['student_id', 'total_score']], on='student_id', how='left')
overview_table = overview_table.drop_duplicates(subset=['student_id'])
# max auto score for tasks 2.1-2.7 is 14 (2 points per task), so max total score without these tasks is 100-14=86
overview_table = overview_table.rename(columns={"total_score": "total_exam_score_without_CBM_out_of_86_points"})
print('Max exam score without CBM tasks is 86')
```
