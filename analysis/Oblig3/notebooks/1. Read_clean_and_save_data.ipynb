## Prepare raw data til analysis

### Oblig 3 2025

```python
df_oblig3 = pd.read_excel('../data/raw_data/oblig3-Iuliia.xlsx', engine='openpyxl')
```

```python
df_oblig3.head()
```

```python
df_oblig3.info()
```

```python
df_oblig3 = df_oblig3.drop(columns=['Brukernavn'])
```

```python
# Remove rows where the consent column is 'Nei´]
df_oblig3 = df_oblig3[df_oblig3['Gir du oss samtykke til å kunne se hvordan dine svar her er i forhold til eksamensresultatet ditt?'] == 'Ja']
anonymize_IDs(df_oblig3, '$submission_id')
```

```python
df_oblig3 = df_oblig3.rename(columns={'$submission_id' : 'kandidatnr'})
df_oblig3 = df_oblig3.rename(columns={'$created' : 'created'})
df_oblig3 = df_oblig3.rename(columns={'Vil du oppgi ditt kjønn' : 'Kjønn'})
df_oblig3['Kjønn'] = df_oblig3['Kjønn'].replace({'Ønsker ikke å oppgi': 'Ukjent'})
df_oblig3 = df_oblig3.rename(columns={'Hvilke studieprogram går du på?' : 'Studieprogram'})
df_oblig3['Studieprogram'] = df_oblig3['Studieprogram'].replace({'Ønsker ikke å oppgi': 'Ukjent'})
df_oblig3 = df_oblig3.rename(columns={'Gir du oss samtykke til å kunne se hvordan dine svar her er i forhold til eksamensresultatet ditt?' : 'Samtykke'})
df_oblig3 = df_oblig3.rename(columns={'$answer_time_ms' : 'Answer_time_ms'})
df_oblig3 = df_oblig3.rename(columns={'DL´ &#43; VN' : 'DL´ + VN'})
df_oblig3 = df_oblig3.rename(columns={'(D&#43;L)(V&#43;N)' : '(D + L)(V + N)'})
df_oblig3 = df_oblig3.rename(columns={'DV(L´ &#43; N)' : 'DV(L´ + N)'})
df_oblig3 = df_oblig3.rename(columns={'D &#43; L &#43; V &#43; N' : 'D + L + V + N'})
```

```python
df_oblig3.head()
```

```python
save_csv_file(df_oblig3, 'oblig3', '../data/cleaned_csv_files/')
```
